# Stage 5: Official SOAP vs Direct Inverse-Root (Decomposition Audit)

This notebook isolates the mechanism behind the stability of the official SOAP implementation. It directly compares the official Adam-style pipeline (`eigenbasis projection` $\rightarrow$ `Adam tracking` $\rightarrow$ `normalization` $\rightarrow$ `inverse projection`) against the direct mathematical `inverse-root` update ($P(G) = A^{-1/2}GB^{-1/2}$), while keeping the first-step skip, EMA logic, and refresh staleness identical.

It also logs the step-by-step magnitude decomposition (G $\rightarrow$ QG $\rightarrow$ m $\rightarrow$ m/sqrt(v) $\rightarrow$ update) and the quantitative staleness of the cached $Q$ basis.

```python
import json
import math
from pathlib import Path
from typing import Dict, List, Optional

import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
class InstrumentedSOAPDecomp(optim.Optimizer):
    """
    Unified optimizer for Stage 5 to compare:
    'official' : Adam in the eigenbasis
    'direct_inverse_root' : P(G) = (A+lambda)^{-1/2} G (B+lambda)^{-1/2}
    """
    def __init__(
        self,
        params,
        lr: float = 1e-3,
        betas=(0.95, 0.95),
        damping: float = 1e-3,
        eps: float = 1e-8,
        precondition_frequency: int=10,
        update_mode: str = "official",
    ):
        defaults = {
            "lr": lr,
            "betas": betas,
            "damping": damping,
            "eps": eps,
            "precondition_frequency": precondition_frequency,
            "update_mode": update_mode,
        }
        super().__init__(params, defaults)
        self.layer_diagnostics = []

    @torch.no_grad()
    def step(self, closure=None):
        loss = None if closure is None else closure()
        self.layer_diagnostics = []
        
        for group in self.param_groups:
            update_mode = group["update_mode"]
            frequency = group["precondition_frequency"]
            damping = group["damping"]
            
            for layer_idx, p in enumerate(group["params"]):
                if p.grad is None:
                    continue
                
                grad = p.grad
                state = self.state[p]
                
                if "step" not in state:
                    state["step"] = 0
                
                if "exp_avg" not in state:
                    state["exp_avg"] = torch.zeros_like(grad)
                    state["exp_avg_sq"] = torch.zeros_like(grad)
                    
                if 'GG' not in state:
                    state['GG'] = []
                    if grad.dim() == 1:
                        state['GG'].append(torch.zeros(grad.shape[0], grad.shape[0], device=grad.device))
                    else:
                        for sh in grad.shape:
                            state['GG'].append(torch.zeros(sh, sh, device=grad.device))
                    
                    state['Q'] = None
                    state['inv_sqrt'] = None
                    
                    self._update_gg(grad, state, group['betas'][1])
                    state['Q'] = self._get_orthogonal_matrix(state['GG'])
                    if update_mode == "direct_inverse_root":
                        state['inv_sqrt'] = self._get_inverse_root(state['GG'], damping)
                    
                    self.layer_diagnostics.append({"layer": layer_idx, "skipped": True})
                    continue # Skip first step 
                
                # Quantify Q Staleness
                Q_fresh = self._get_orthogonal_matrix(state['GG'])
                E_Q_list = []
                if state['Q'] is not None:
                    for q_u, q_f in zip(state['Q'], Q_fresh):
                        if len(q_u) > 0 and len(q_f) > 0:
                            proj_u = q_u @ q_u.T
                            proj_f = q_f @ q_f.T
                            E_Q_list.append(float(torch.linalg.matrix_norm(proj_u - proj_f, 'fro')))
                E_Q = max(E_Q_list) if len(E_Q_list) > 0 else 0.0
                
                if update_mode == "official":
                    norm_G = float(torch.linalg.vector_norm(grad))
                    Q_G = self._project(grad, state['Q'])
                    norm_QG = float(torch.linalg.vector_norm(Q_G))
                    
                    exp_avg, exp_avg_sq = state["exp_avg"], state["exp_avg_sq"]
                    beta1, beta2 = group["betas"]
                    state["step"] += 1
                    
                    exp_avg.mul_(beta1).add_(Q_G, alpha=(1.0 - beta1))
                    exp_avg_sq.mul_(beta2).add_(Q_G.square(), alpha=(1.0 - beta2))
                    
                    denom = exp_avg_sq.sqrt().add_(group["eps"])
                    m_norm = float(torch.linalg.vector_norm(exp_avg))
                    
                    normalized_update = exp_avg / denom
                    norm_normalized = float(torch.linalg.vector_norm(normalized_update))
                    
                    step_size = group["lr"]
                    bias_correction1 = 1.0 - beta1 ** state["step"]
                    bias_correction2 = 1.0 - beta2 ** state["step"]
                    step_size = step_size * (bias_correction2 ** .5) / bias_correction1
                    
                    norm_grad = self._project_back(normalized_update, state['Q'])
                    update_norm = float(torch.linalg.vector_norm(norm_grad))
                    
                    p.add_(norm_grad, alpha=-step_size)
                    
                    self.layer_diagnostics.append({
                        "layer": layer_idx,
                        "skipped": False,
                        "E_Q": E_Q,
                        "norm_G": norm_G,
                        "norm_QG": norm_QG,
                        "norm_m": m_norm,
                        "norm_normalized": norm_normalized,
                        "update_norm": update_norm,
                        "precond_gain": update_norm / (norm_G + 1e-12)
                    })
                    
                else:
                    norm_G = float(torch.linalg.vector_norm(grad))
                    state["step"] += 1
                    
                    norm_grad = self._apply_inverse_root(grad, state['inv_sqrt'])
                    update_norm = float(torch.linalg.vector_norm(norm_grad))
                    
                    p.add_(norm_grad, alpha=-group["lr"])
                    
                    self.layer_diagnostics.append({
                        "layer": layer_idx,
                        "skipped": False,
                        "E_Q": E_Q,
                        "norm_G": norm_G,
                        "update_norm": update_norm,
                        "precond_gain": update_norm / (norm_G + 1e-12)
                    })
                
                if update_mode == "official" and state["Q"] is not None:
                    state["exp_avg"] = self._project_back(state["exp_avg"], state["Q"])
                    
                self._update_gg(grad, state, group['betas'][1])
                
                if state['Q'] is None or (state['step'] > 0 and state['step'] % frequency == 0):
                    state['Q'] = self._get_orthogonal_matrix(state['GG'])
                    if update_mode == "direct_inverse_root":
                        state['inv_sqrt'] = self._get_inverse_root(state['GG'], damping)
                        
                if update_mode == "official" and state["step"] > 0:
                    state["exp_avg"] = self._project(state["exp_avg"], state["Q"])
                    
        return loss

    def _update_gg(self, grad, state, beta2):
        if grad.dim() == 1:
            state['GG'][0].lerp_(grad.unsqueeze(1) @ grad.unsqueeze(0), 1 - beta2)
        else:
            state['GG'][0].lerp_(torch.tensordot(grad, grad, dims=[[1], [1]]), 1 - beta2)
            state['GG'][1].lerp_(torch.tensordot(grad, grad, dims=[[0], [0]]), 1 - beta2)

    def _project(self, grad, Q_list):
        for mat in Q_list:
            if len(mat) > 0:
                grad = torch.tensordot(grad, mat, dims=[[0], [0]])
            else:
                permute_order = list(range(1, len(grad.shape))) + [0]
                grad = grad.permute(permute_order)
        return grad

    def _project_back(self, grad, Q_list):
        for mat in Q_list:
            if len(mat) > 0:
                grad = torch.tensordot(grad, mat, dims=[[0], [1]])
            else:
                permute_order = list(range(1, len(grad.shape))) + [0]
                grad = grad.permute(permute_order)
        return grad

    def _get_orthogonal_matrix(self, mat_list):
        final = []
        for m in mat_list:
            if len(m) == 0:
                final.append([])
                continue
            m64 = m.to(torch.float64)
            m64 = 0.5 * (m64 + m64.T)
            _, Q = torch.linalg.eigh(m64 + 1e-30*torch.eye(m.shape[0], device=m.device))
            Q = torch.flip(Q, [1])
            final.append(Q.to(m.dtype))
        return final

    def _get_inverse_root(self, mat_list, damping):
        final = []
        for m in mat_list:
            if len(m) == 0:
                final.append([])
                continue
            m64 = m.to(torch.float64)
            m64 = 0.5 * (m64 + m64.T)
            eigenvalues, eigenvectors = torch.linalg.eigh(m64 + damping*torch.eye(m.shape[0], device=m.device))
            eigenvalues = torch.clamp(eigenvalues, min=damping)
            inv_sqrt = torch.rsqrt(eigenvalues)
            res = eigenvectors @ torch.diag(inv_sqrt) @ eigenvectors.T
            final.append(res.to(m.dtype))
        return final

    def _apply_inverse_root(self, grad, inv_sqrt_list):
        if grad.dim() == 1:
            if len(inv_sqrt_list[0]) > 0:
                return inv_sqrt_list[0] @ grad
            return grad
        else:
            res = grad
            if len(inv_sqrt_list[0]) > 0:
                res = inv_sqrt_list[0] @ res
            if len(inv_sqrt_list[1]) > 0:
                res = res @ inv_sqrt_list[1]
            return res
```

```python
class PINN(nn.Module):
    def __init__(self, width: int = 128, hidden_layers: int = 4) -> None:
        super().__init__()
        layers: List[nn.Module] = [nn.Linear(2, width), nn.Tanh()]
        for _ in range(hidden_layers - 1):
            layers.extend([nn.Linear(width, width), nn.Tanh()])
        layers.append(nn.Linear(width, 1))
        self.net = nn.Sequential(*layers)
        self.reset_parameters()

    def reset_parameters(self) -> None:
        for module in self.modules():
            if isinstance(module, nn.Linear):
                nn.init.xavier_uniform_(module.weight)
                nn.init.zeros_(module.bias)

    def forward(self, xy: torch.Tensor) -> torch.Tensor:
        return self.net(xy)

def set_seed(seed: int) -> None:
    torch.manual_seed(seed)
    np.random.seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
```

```python
def sample_collocation(n: int, device: torch.device) -> torch.Tensor:
    return torch.rand(n, 2, device=device, requires_grad=True)

def sample_boundary(n_each: int, device: torch.device) -> torch.Tensor:
    t = torch.rand(4 * n_each, device=device)
    x = torch.empty(4 * n_each, 2, device=device)
    x[:n_each] = torch.stack([torch.zeros(n_each, device=device), t[:n_each]], dim=1)
    x[n_each:2*n_each] = torch.stack([torch.ones(n_each, device=device), t[n_each:2*n_each]], dim=1)
    x[2*n_each:3*n_each] = torch.stack([t[2*n_each:3*n_each], torch.zeros(n_each, device=device)], dim=1)
    x[3*n_each:] = torch.stack([t[3*n_each:], torch.ones(n_each, device=device)], dim=1)
    return x

def make_batch_schedule(steps: int, K: int, n_pde: int, n_boundary_each: int, device: torch.device) -> List[Dict[str, torch.Tensor]]:
    schedule: List[Dict[str, torch.Tensor]] = []
    pde: Optional[torch.Tensor] = None
    bc: Optional[torch.Tensor] = None

    for step in range(1, steps + 1):
        if step == 1 or (step - 1) % K == 0:
            pde = sample_collocation(n_pde, device)
            bc = sample_boundary(n_boundary_each, device)
        schedule.append({
            "pde": pde.detach().clone(),
            "bc": bc.detach().clone(),
        })
    return schedule
```

```python
def poisson_residual(model: nn.Module, xy: torch.Tensor) -> torch.Tensor:
    u = model(xy)
    grad_u = torch.autograd.grad(u, xy, grad_outputs=torch.ones_like(u), create_graph=True, retain_graph=True)[0]
    u_x, u_y = grad_u[:, 0:1], grad_u[:, 1:2]
    u_xx = torch.autograd.grad(u_x, xy, grad_outputs=torch.ones_like(u_x), create_graph=True, retain_graph=True)[0][:, 0:1]
    u_yy = torch.autograd.grad(u_y, xy, grad_outputs=torch.ones_like(u_y), create_graph=True, retain_graph=True)[0][:, 1:2]
    forcing = 2.0 * math.pi**2 * torch.sin(math.pi * xy[:, 0:1]) * torch.sin(math.pi * xy[:, 1:2])
    return -(u_xx + u_yy) - forcing

def relative_l2(model: nn.Module, device: torch.device, n: int = 81) -> float:
    grid = torch.linspace(0.0, 1.0, n, device=device)
    xx, yy = torch.meshgrid(grid, grid, indexing="ij")
    xy = torch.stack([xx.reshape(-1), yy.reshape(-1)], dim=1)
    with torch.no_grad():
        pred = model(xy).reshape(-1)
    ref = torch.sin(math.pi * xy[:, 0]) * torch.sin(math.pi * xy[:, 1])
    return float(torch.linalg.vector_norm(pred - ref) / (torch.linalg.vector_norm(ref) + 1e-12))
```

```python
def run_stage5_audit(update_mode: str, K: int=1, steps: int=20, seed: int=0):
    set_seed(seed)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = PINN().to(device)
    
    schedule_path = Path(f"poisson_seed{seed}_K{K}_schedule.pt")
    if schedule_path.exists():
        schedule = torch.load(schedule_path, map_location=device, weights_only=True)
    else:
        schedule = make_batch_schedule(max(2000, steps), K, 1000, 100, device)
        torch.save(schedule, schedule_path)

    opt = InstrumentedSOAPDecomp(model.parameters(), lr=1e-3, betas=(0.95, 0.95), precondition_frequency=10, update_mode=update_mode)

    history = []
    model.train()
    for step in range(1, steps + 1):
        current_pde = schedule[step - 1]["pde"].to(device).detach().clone().requires_grad_(True)
        current_bc = schedule[step - 1]["bc"].to(device)
        opt.zero_grad(set_to_none=True)

        r = poisson_residual(model, current_pde)
        loss_pde = torch.mean(r.square())
        u_bc = model(current_bc)
        loss_bc = torch.mean(u_bc.square())
        loss = loss_pde + loss_bc
        loss.backward()

        opt.step()
        
        err = relative_l2(model, device)
        
        step_data = {
            "step": step,
            "loss": float(loss.detach()),
            "rel_l2": float(err),
            "layers": opt.layer_diagnostics
        }
        history.append(step_data)
        
        print(f"Step {step} | Loss {loss.item():.4e} | RelL2 {err:.4e}")

    return history
```

### Stage 5A: Official SOAP vs Direct Inverse-Root (K=1, 20 steps)

```python
print("Running K=1 Official SOAP Pipeline...")
audit_official = run_stage5_audit(update_mode="official", K=1, steps=20)
with open("stage5_official_pipeline.json", "w") as f:
    json.dump(audit_official, f, indent=2)
```

```python
print("\nRunning K=1 Direct Inverse-Root (Matched stale setup)....")
audit_direct = run_stage5_audit(update_mode="direct_inverse_root", K=1, steps=20)
with open("stage5_direct_inverse_root.json", "w") as f:
    json.dump(audit_direct, f, indent=2)
```
